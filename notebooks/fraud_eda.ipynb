```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv("../data/raw/creditcard.csv")
```

```python
data.head()
```

```python
data.Class.value_counts(dropna= False)
```

```python
# Check class balance
data['Class'].value_counts(normalize=True) * 100
```

Insight: Expect to see 0.17% fraud cases (heavily imbalanced dataset).

```python
sns.countplot(x = 'Class' , data = data)
plt.title("Class Distribution (Legit(0) vs Fraud(1))")
plt.show()
```

Insight: Fraud is extremely rare → needs resampling or anomaly detection later.

```python
# Count the values
class_counts = data['Class'].value_counts()

# Pie chart
plt.figure(figsize=(6,6))
plt.pie(class_counts,
        labels=class_counts.index,
        autopct='%1.1f%%',   # show percentages
        startangle=90,       # rotate to start at the top
        counterclock=False)  # clockwise
plt.title("Class Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(data["Amount"],bins = 100 , log_scale=True , kde= False)
plt.title("Transaction Amount Distribution")
plt.show()

# Fraud vs Legit Amount comparison
plt.figure(figsize=(8,4))
sns.boxplot(x="Class", y="Amount", data=data)
plt.title("Transaction Amount by Class")
plt.show()
```

Insight: Fraud amounts may differ (often smaller to avoid detection).

```python
corr = data.corr()

plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()

# Focus correlation with Class
corr["Class"].sort_values(ascending=False)
```
